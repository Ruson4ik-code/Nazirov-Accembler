## 2. Вычислительные задачи

1. Разработать программу на языке ассемблера, которая принимает в качестве параметра командной строки значение $2n+1$, выполняет расчет указанного выражения и выводит результат на экран:
$$1+3+5+7+\ldots +(2n+1)$$
Если пользователь ввел четное число, программа должна выдать ошибку.
2. Разработать программу на языке ассемблера, которая принимает в качестве параметра командной строки значение $n$, выполняет расчет указанного выражения и выводит результат на экран:
$$1+11+111+1111+\ldots+\sum_{k=0}^n 10^k$$
3. Разработать программу на языке ассемблера, которая принимает в качестве параметра командной строки значение $n$, выполняет расчет указанного выражения и выводит результат на экран:
$$1+2+\ldots+11+21+31+\ldots+91+2+12+\ldots$$
Т.е. суммируются реверсированные числа от 1 до $n$.
4. Разработать программу на языке ассемблера, которая принимает два параметра командной строки $n, k$, суммирует все числа от 1 до $n$, у которых сумма цифр равна $k$.
5. Разработать программу на языке ассемблера, которая принимает в качестве параметра командной строки значение $n$, выполняет расчет указанного выражения и выводит результат на экран:
$$\sum_{k=1}^n k\cdot first\_digit(k) $$
где $first\_digit(k)$ - первая значащая цифра числа  $k$. Например, $first\_digit(56743) = 5$.
